## 1.1 Generate search queries based on the project description


```python
from pubmed import PubMedAPI

api = PubMedAPI()
papers = api.search("ovarian aging oocyte quality", max_results=3, meta_analysis_only=True)
for i, paper in enumerate(papers, 1):
    print(f"\n{i}. {paper['title']}")
    print(f"   Authors: {', '.join(paper['authors'][:3])}")
    print(f"   Journal: {paper['journal']} ({paper['pub_date']})")
    print(f"   DOI: {paper['doi']}")
    print(f"   PMID: {paper['pmid']}")
    print(f"   Abstract: {paper['abstract'][:150]}..." if len(paper['abstract']) > 150 else f"   Abstract: {paper['abstract']}")
```
